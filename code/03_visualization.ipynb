## Confusion Matrix Analysis

```python
def analyze_labels(df, evaluation_col, true_label_col, sample_size='all', min_word_count=None,
                   exclude_roles=None, exclude_topics=None, text_column='Text_English', 
                   plot_size=(16, 12), dataset_name="Dataset"):
    """Generate confusion matrix and metrics"""
    
    exclude_roles = exclude_roles or []
    exclude_topics = exclude_topics or []
    
    # Filter
    df = df[~df[true_label_col].isin(['-'])].dropna(subset=[evaluation_col, true_label_col])
    
    if exclude_roles:
        df = df[~df['Speaker_role'].isin(exclude_roles)]
    if exclude_topics:
        df = df[~df[evaluation_col].isin(exclude_topics)]
        df = df[~df[true_label_col].isin(exclude_topics)]
    if min_word_count:
        df = df[df[text_column].apply(lambda x: len(str(x).split())) >= min_word_count]
    
    # Sample
    if sample_size != 'all':
        sampled = []
        for cat in df[true_label_col].unique():
            cat_df = df[df[true_label_col] == cat]
            sampled.append(cat_df.sample(n=min(sample_size, len(cat_df))))
        df = pd.concat(sampled, ignore_index=True)
    
    # Metrics
    y_true = df[true_label_col]
    y_pred = df[evaluation_col]
    f1_macro = f1_score(y_true, y_pred, average='macro', zero_division=0)
    f1_micro = f1_score(y_true, y_pred, average='micro', zero_division=0)
    
    # Plot
    plt.figure(figsize=plot_size)
    categories = y_true.value_counts().index.tolist()
    conf_matrix = confusion_matrix(y_true, y_pred, labels=categories)
    sns.heatmap(conf_matrix, xticklabels=categories, yticklabels=categories, annot=True, 
                fmt='d', cmap='Blues', square=True, annot_kws={'size': 8})
    plt.title(f'{dataset_name}: F1 macro={f1_macro:.2f}, micro={f1_micro:.2f}')
    plt.xlabel('Predicted')
    plt.ylabel('True Label')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.savefig(os.path.join(confusion_dir, f'confusion_{dataset_name.lower()}.png'), dpi=300, bbox_inches='tight')
    plt.show()
    
    return {'f1_macro': f1_macro, 'f1_micro': f1_micro, 'total': len(df)}

# Run analysis
results = {}
results['GB'] = analyze_labels(GB, 'topic_consensus', 'True_label', exclude_topics=['Mix'], dataset_name="GB")
results['HR'] = analyze_labels(HR, 'topic_consensus', 'True_label', exclude_topics=['Mix'], dataset_name="HR")
results['AT'] = analyze_labels(AT, 'topic_consensus', 'Topic', exclude_topics=['Mix'], 
                               min_word_count=70, dataset_name="AT")
```

## Topic Distribution Comparison

```python
def compare_topic_distributions(df, our_col, comparison_col, dataset_name, exclude_topics=['Other', 'Mix']):
    """Compare topic distributions"""
    clean = df.dropna(subset=[our_col, comparison_col])
    for topic in exclude_topics:
        clean = clean[(clean[our_col] != topic) & (clean[comparison_col] != topic)]
    
    # Check if we have enough data after filtering
    if len(clean) == 0:
        print(f"⚠️ Warning: {dataset_name} has no overlapping topics after filtering {exclude_topics}")
        return
    
    our_dist = clean[our_col].value_counts(normalize=True) * 100
    comp_dist = clean[comparison_col].value_counts(normalize=True) * 100
    
    # Check if we have any topics to compare
    if len(our_dist) == 0 or len(comp_dist) == 0:
        print(f"⚠️ Warning: {dataset_name} has no topics to compare after filtering")
        return
    
    comparison = pd.DataFrame({'Our': our_dist, 'Reference': comp_dist}).fillna(0)
    
    # Check if comparison has any rows
    if len(comparison) == 0:
        print(f"⚠️ Warning: {dataset_name} comparison dataframe is empty")
        return
    
    diff = (comparison['Our'] - comparison['Reference']).reindex(
        (comparison['Our'] - comparison['Reference']).abs().sort_values(ascending=True).index)
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20, 8))
    
    comparison.plot(kind='barh', ax=ax1, width=0.8, alpha=0.8)
    ax1.set_title(f'{dataset_name}: Topic Distribution Comparison')
    ax1.set_xlabel('Percentage (%)')
    ax1.grid(axis='x', alpha=0.3)
    
    colors = ['red' if x < 0 else 'green' for x in diff]
    diff.plot(kind='barh', ax=ax2, color=colors, alpha=0.7)
    ax2.set_title(f'{dataset_name}: Difference (Our - Reference)')
    ax2.set_xlabel('Percentage Point Difference')
    ax2.axvline(x=0, color='black', linestyle='-', alpha=0.3)
    ax2.grid(axis='x', alpha=0.3)
    
    plt.tight_layout()
    plt.savefig(os.path.join(topic_dist_dir, f'dist_{dataset_name.lower()}.png'), dpi=300, bbox_inches='tight')
    plt.show()

compare_topic_distributions(GB, 'topic_consensus', 'True_label', 'GB')
compare_topic_distributions(HR, 'topic_consensus', 'True_label', 'HR')
compare_topic_distributions(AT, 'topic_consensus', 'Topic', 'AT')

print("✅ Topic classification evaluation complete")
```

# PART 2: LIWC Linguistic Analysis

Analyzing political discourse using LIWC-22 dimensions across countries, topics, and covariates.

```python
# Load LIWC benchmarks
LIWC_statistics_path = os.path.join("data folder", "LIWC-22.Descriptive.Statistics-Test.Kitchen.xlsx")

def load_liwc_benchmarks(file_path):
    """Load LIWC-22 population norms"""
    raw = pd.read_excel(file_path, sheet_name=0, header=None)
    header_row = raw.iloc[0]
    total_col_start = [i for i, v in enumerate(header_row) if str(v).strip() == 'Total'][0]
    
    dimensions = raw.iloc[2:, 0].dropna().reset_index(drop=True)
    means = pd.to_numeric(raw.iloc[2:2+len(dimensions), total_col_start], errors='coerce')
    stds = pd.to_numeric(raw.iloc[2:2+len(dimensions), total_col_start+1], errors='coerce')
    
    return pd.DataFrame({'Dimension': dimensions, 'Mean': means.values, 'Std': stds.values}).dropna()

LIWC_benchmarks = load_liwc_benchmarks(LIWC_statistics_path)

# Combine datasets for analysis
LIWC_ALL = pd.concat([AT, HR, GB], ignore_index=True)

# Rename for consistency
LIWC_ALL.rename(columns={'topic_consensus': 'Our_Topic', 'Topic': 'ParlaCAP'}, inplace=True)

# Filter out non-policy content
LIWC_ALL = LIWC_ALL[~LIWC_ALL['Our_Topic'].isin(['Mix', 'Other'])]
LIWC_ALL = LIWC_ALL[LIWC_ALL['Speaker_role'] != 'Chairperson']

# Key dimensions
KEY_LIWC_DIMENSIONS = [
    'Analytic', 'Clout', 'Authentic', 'Tone',
    'i', 'we', 'you', 'they', 'ipron', 'ppron',
    'focuspast', 'focuspresent', 'focusfuture',
    'cogproc', 'insight', 'cause', 'discrep', 'tentat', 'certitude',
    'Affect', 'tone_pos', 'tone_neg',
    'Social', 'conflict', 'moral',
    'power', 'politic', 'money', 'work'
]

print(f"✅ LIWC analysis ready: {len(LIWC_ALL):,} speeches")
print(f"   Countries: {LIWC_ALL['Country'].unique().tolist()}")
print(f"   Key dimensions: {len(KEY_LIWC_DIMENSIONS)}")
```

```python
def create_country_heatmap(data, benchmarks, dimensions):
    """Z-score heatmap with categorical groupings and labels"""
    country_means = data.groupby('Country')[dimensions].mean()
    overall_means = data[dimensions].mean()
    country_means.loc['All Countries'] = overall_means
    
    benchmark_lookup = benchmarks.set_index('Dimension')
    z_scores = (country_means - benchmark_lookup.loc[dimensions, 'Mean']) / benchmark_lookup.loc[dimensions, 'Std']
    
    # Define groups exactly as in your image
    summary_vars = ['Analytic', 'Clout', 'Authentic', 'Tone']
    
    # Personal Pronouns group
    pronoun_vars = ['i', 'we', 'you', 'shehe', 'they', 'ipron', 'ppron']
    pronoun_vars = [v for v in pronoun_vars if v in dimensions]
    
    # Time Orientation group
    temporal_vars = ['focuspast', 'focuspresent', 'focusfuture']
    temporal_vars = [v for v in temporal_vars if v in dimensions]
    
    # Cognitive Processes group
    cognitive_vars = ['cogproc', 'insight', 'cause', 'discrep', 'tentat', 'certitude']
    cognitive_vars = [v for v in cognitive_vars if v in dimensions]
    
    # Emotion group
    affect_vars = ['Affect', 'tone_pos', 'tone_neg']
    affect_vars = [v for v in affect_vars if v in dimensions]
    
    # Social group
    social_vars = ['Social', 'conflict', 'moral']
    social_vars = [v for v in social_vars if v in dimensions]
    
    # Power & Politics group
    political_vars = ['power', 'politic', 'money', 'work']
    political_vars = [v for v in political_vars if v in dimensions]
    
    # Build ordered list
    ordered_dims = []
    category_labels = []
    separator_positions = []
    
    # Add Summary Variables
    ordered_dims.extend(summary_vars)
    category_labels.append(('Summary\nVariables', 0, len(summary_vars)))
    separator_positions.append(len(ordered_dims))
    
    # Add Personal Pronouns
    if pronoun_vars:
        start_pos = len(ordered_dims)
        ordered_dims.extend(pronoun_vars)
        category_labels.append(('Personal\nPronouns', start_pos, len(ordered_dims)))
        separator_positions.append(len(ordered_dims))
    
    # Add Time Orientation
    if temporal_vars:
        start_pos = len(ordered_dims)
        ordered_dims.extend(temporal_vars)
        category_labels.append(('Time\nOrientation', start_pos, len(ordered_dims)))
        separator_positions.append(len(ordered_dims))
    
    # Add Cognitive Processes
    if cognitive_vars:
        start_pos = len(ordered_dims)
        ordered_dims.extend(cognitive_vars)
        category_labels.append(('Cognitive\nProcesses', start_pos, len(ordered_dims)))
        separator_positions.append(len(ordered_dims))
    
    # Add Emotion
    if affect_vars:
        start_pos = len(ordered_dims)
        ordered_dims.extend(affect_vars)
        category_labels.append(('Emotion', start_pos, len(ordered_dims)))
        separator_positions.append(len(ordered_dims))
    
    # Add Social
    if social_vars:
        start_pos = len(ordered_dims)
        ordered_dims.extend(social_vars)
        category_labels.append(('Social', start_pos, len(ordered_dims)))
        separator_positions.append(len(ordered_dims))
    
    # Add Power & Politics
    if political_vars:
        start_pos = len(ordered_dims)
        ordered_dims.extend(political_vars)
        category_labels.append(('Power\n&\nPolitics', start_pos, len(ordered_dims)))
        separator_positions.append(len(ordered_dims))
    
    z_scores_ordered = z_scores[ordered_dims].T
    
    fig, ax = plt.subplots(figsize=(8, 14))
    sns.heatmap(z_scores_ordered, annot=True, fmt=".2f", cmap="RdBu_r", center=0, 
                cbar_kws={'label': 'Z-Score (from population norm)'}, vmin=-2, vmax=2, ax=ax, 
                square=False, annot_kws={'size': 9, 'color': 'black'})
    
    # Add separator lines
    for pos in separator_positions[:-1]:
        ax.axhline(y=pos, color='black', linewidth=2)
    
    # Add category labels on the left - moved further left
    for label_text, start, end in category_labels:
        y_pos = (start + end) / 2
        ax.text(-1.2, y_pos, label_text, 
                fontsize=11, fontweight='bold', color='darkblue',
                verticalalignment='center', horizontalalignment='right')
    
    ax.set_title("LIWC Z-Scores vs. Population Norms\n(All KEY_LIWC_DIMENSIONS)", 
                 fontweight='bold', fontsize=12)
    plt.setp(ax.get_xticklabels(), rotation=45, ha='right')
    plt.tight_layout()
    plt.savefig(os.path.join(liwc_dir, 'z_scores_countries.png'), dpi=300, bbox_inches='tight')
    plt.show()

create_country_heatmap(LIWC_ALL, LIWC_benchmarks, KEY_LIWC_DIMENSIONS)
```

## Topic-LIWC Interaction Analysis

```python
def create_topic_liwc_heatmap(data, benchmarks, topic_col='Our_Topic', 
                               top_n_topics=8):
    """Create topic-LIWC interaction heatmap for each country as a separate figure."""
    
    # Fixed dimensions (always first 5)
    fixed_dims = ['Analytic', 'Clout', 'Authentic', 'Tone', 'power']
    
    # Get top topics by frequency
    top_topics = data[topic_col].value_counts().head(top_n_topics).index.tolist()
    filtered_data = data[data[topic_col].isin(top_topics)].copy()
    
    # Z-score normalize all KEY_LIWC_DIMENSIONS
    benchmark_lookup = benchmarks.set_index('Dimension')
    for dim in KEY_LIWC_DIMENSIONS:
        if dim in benchmark_lookup.index:
            mean = benchmark_lookup.loc[dim, 'Mean']
            std = benchmark_lookup.loc[dim, 'Std']
            filtered_data[f'{dim}_z'] = (filtered_data[dim] - mean) / std
    
    countries = ['Austria', 'Croatia', 'Great Britain']
    
    for country in countries:
        fig, ax = plt.subplots(figsize=(16, 8)) # Create a new figure for each country
        
        country_data = filtered_data[filtered_data['Country'] == country]
        
        # Calculate z-scores by topic for all dimensions
        z_cols = [f'{d}_z' for d in KEY_LIWC_DIMENSIONS if f'{d}_z' in country_data.columns]
        topic_means = country_data.groupby(topic_col)[z_cols].mean()
        
        # Calculate average absolute z-score for each dimension (excluding fixed)
        other_cols = [c for c in topic_means.columns if c.replace('_z', '') not in fixed_dims]
        avg_abs_z = topic_means[other_cols].abs().mean(axis=0).sort_values(ascending=False)
        top_5_other = avg_abs_z.head(5).index.tolist()
        
        # Combine fixed + top 5 by z-score
        selected_cols = [f'{d}_z' for d in fixed_dims] + top_5_other
        topic_means_selected = topic_means[selected_cols]
        topic_means_selected.columns = [c.replace('_z', '') for c in topic_means_selected.columns]
        
        # Order topics by frequency
        topic_order = country_data[topic_col].value_counts().index
        topic_means_selected = topic_means_selected.reindex(topic_order)
        
        # Transpose for dimensions as rows
        topic_means_T = topic_means_selected.T
        
        # Plot
        sns.heatmap(topic_means_T, annot=True, fmt='.2f', cmap='RdBu_r', 
                   center=0, vmin=-1.5, vmax=1.5, ax=ax, 
                   cbar_kws={'label': 'Z-Score'}, square=False,
                   annot_kws={'color': 'black'})
        
        # Add separator line after first 5 dimensions
        ax.axhline(y=5, color='black', linewidth=2)
        
        ax.set_title(f'Topic-LIWC Interaction: {country}', fontweight='bold', fontsize=16)
        ax.set_xlabel('Policy Topic', fontweight='bold')
        ax.set_ylabel('LIWC Dimension', fontweight='bold')
        
        plt.setp(ax.get_xticklabels(), rotation=45, ha='right')
    
        plt.tight_layout()
        
        # Save each figure separately
        filename = f'topic_liwc_interaction_{country.lower().replace(" ", "_")}.png'
        plt.savefig(os.path.join(liwc_dir, filename), 
                    dpi=300, bbox_inches='tight')
        plt.show()

create_topic_liwc_heatmap(LIWC_ALL, LIWC_benchmarks)
      
```

```python
def create_focal_topic_analysis(data, benchmarks, focal_topics):
    """Compare focal topics vs all other topics with fixed top 5 and topic-specific bottom 5"""
    
    # Fixed dimensions (always first 5)
    fixed_dims = ['Analytic', 'Authentic', 'Clout', 'Tone', 'power']
    
    countries = ['Austria', 'Croatia', 'Great Britain', 'All Countries']
    
    # Calculate results for each topic independently
    results = {}
    topic_top5 = {}  # Store top 5 for each topic
    
    for focal_topic in focal_topics:
        topic_results = []
        all_diffs = []  # Track all differences for this topic
        
        for country in countries:
            if country == 'All Countries':
                country_data = data.copy()
            else:
                country_data = data[data['Country'] == country].copy()
            
            focal_data = country_data[country_data['Our_Topic'] == focal_topic]
            other_data = country_data[country_data['Our_Topic'] != focal_topic]
            
            benchmark_lookup = benchmarks.set_index('Dimension')
            diffs = {}
            for dim in KEY_LIWC_DIMENSIONS:
                if dim in benchmark_lookup.index:
                    mean = benchmark_lookup.loc[dim, 'Mean']
                    std = benchmark_lookup.loc[dim, 'Std']
                    
                    focal_z = (focal_data[dim].mean() - mean) / std
                    other_z = (other_data[dim].mean() - mean) / std
                    diff = focal_z - other_z
                    diffs[dim] = diff
                    
                    # Track for top 5 selection (exclude fixed dims)
                    if dim not in fixed_dims:
                        all_diffs.append({'dimension': dim, 'country': country, 'diff': abs(diff)})
            
            topic_results.append(diffs)
        
        # Select top 5 non-fixed dimensions by average absolute difference for THIS topic
        diff_df = pd.DataFrame(all_diffs)
        top_5_other = diff_df.groupby('dimension')['diff'].mean().nlargest(5).index.tolist()
        topic_top5[focal_topic] = top_5_other
        
        # Store results with selected dimensions
        selected_dims = fixed_dims + top_5_other
        results_df = pd.DataFrame(topic_results, index=countries).T
        results[focal_topic] = results_df.loc[selected_dims]
    
    # Create subplots
    n_topics = len(focal_topics)
    fig, axes = plt.subplots(n_topics, 1, figsize=(14, 10 * n_topics))
    if n_topics == 1:
        axes = [axes]
    
    for idx, focal_topic in enumerate(focal_topics):
        ax = axes[idx]
        data_to_plot = results[focal_topic]
        
        im = sns.heatmap(data_to_plot, annot=True, fmt='.2f', cmap='RdBu_r', 
                   center=0, vmin=-1, vmax=1, ax=ax, square=False,
                   cbar_kws={'label': 'Z-Score Difference (Focal Topic - Other)'},
                   annot_kws={'size': 16, 'color': 'black'})
        
        # Set colorbar label size
        cbar = im.collections[0].colorbar
        cbar.ax.tick_params(labelsize=12)
        cbar.set_label('Z-Score Difference (Focal Topic - Other)', size=14)
        
        # Add separator line after first 5 dimensions
        ax.axhline(y=5, color='black', linewidth=2)
        
        ax.set_title(f'{focal_topic} vs Other Topics', fontweight='bold', fontsize=16)
        ax.set_xlabel('', fontsize=14)
        ax.set_ylabel('LIWC Dimension', fontweight='bold', fontsize=14)
        ax.tick_params(axis='both', labelsize=13)
    
    fig.suptitle('Focal Topics Analysis: Linguistic Differences', 
                 fontsize=18, fontweight='bold', y=0.995)
    plt.tight_layout()
    plt.savefig(os.path.join(liwc_dir, 'focal_topics_analysis.png'), 
                dpi=300, bbox_inches='tight')
    plt.show()

create_focal_topic_analysis(LIWC_ALL, LIWC_benchmarks, 
                            focal_topics=['Macroeconomics', 'Health'])
```

## Political Covariates Analysis

```python
def create_covariate_heatmap_with_counts(data, benchmarks, covariate_col, categories, 
                                         dimensions, title, filename, order_by_effect=False):
    """Covariate heatmap with sample sizes in labels"""
    filtered = data[data[covariate_col].isin(categories)].copy()
    
    # Get sample sizes
    sample_sizes = filtered[covariate_col].value_counts()
    
    # Z-score normalize
    benchmark_lookup = benchmarks.set_index('Dimension')
    for dim in dimensions:
        if dim in benchmark_lookup.index:
            mean = benchmark_lookup.loc[dim, 'Mean']
            std = benchmark_lookup.loc[dim, 'Std']
            filtered[f'{dim}_z'] = (filtered[dim] - mean) / std
    
    means = filtered.groupby(covariate_col)[[f'{d}_z' for d in dimensions]].mean()
    means.columns = [c.replace('_z', '') for c in means.columns]
    
    # Order by effect size if requested
    if order_by_effect and len(categories) == 2:
        diff = (means.iloc[0] - means.iloc[1]).abs().sort_values(ascending=False)
        means = means[diff.index]
    
    # Add horizontal line after first group (summary variables + top effect)
    separator_position = None
    if 'power' in means.columns:
        separator_position = list(means.columns).index('power') + 1
    
    fig, ax = plt.subplots(figsize=(10, 8))
    sns.heatmap(means.T, annot=True, fmt='.2f', cmap='RdBu_r', center=0, 
                vmin=-1, vmax=1, ax=ax, square=False,
                cbar_kws={'label': 'Z-Score Diff (Coalition - Opposition)' if 'Coalition' in categories 
                         else 'Z-Score Diff (Female - Male)' if 'F' in categories
                         else 'Z-Score'},
                annot_kws={'color': 'black'})
    
    # Add separator line
    if separator_position:
        ax.axhline(y=separator_position, color='black', linewidth=2)
    
    ax.set_title(title, fontweight='bold')
    
    # Update column labels with sample sizes - 2 lines for 2 categories
    if len(categories) == 2:
        col_labels = [f"{cat}\nN={sample_sizes[cat]:,}" for cat in means.index]
    else:
        col_labels = [f"{cat}\n(N={sample_sizes[cat]:,})" for cat in means.index]
    ax.set_xticklabels(col_labels, rotation=0)
    
    # Add total sample size info at bottom
    total_n = sum(sample_sizes)
    coalition_n = sample_sizes.get('Coalition', 0)
    opp_n = sample_sizes.get('Opposition', 0)
    
    if coalition_n and opp_n:
        bottom_text = f"Coalition={coalition_n:,}\nOpposition={opp_n:,}"
        fig.text(0.5, -0.02, bottom_text, ha='center', fontsize=10)
    elif 'F' in sample_sizes.index and 'M' in sample_sizes.index:
        f_n = sample_sizes['F']
        m_n = sample_sizes['M']
        bottom_text = f"Female={f_n:,}\nMale={m_n:,}"
        fig.text(0.5, -0.02, bottom_text, ha='center', fontsize=10)
    
    plt.tight_layout()
    plt.savefig(os.path.join(liwc_dir, filename), dpi=300, bbox_inches='tight')
    plt.show()

def create_country_split_heatmap(data, benchmarks, covariate_col, categories, 
                                 dimensions, title, filename):
    """Create side-by-side heatmaps for each country with 10 dimensions (5 fixed + top 5 by effect)"""
    countries = ['Austria', 'Croatia', 'Great Britain', 'All Countries']
    
    # Fixed dimensions (always first 5)
    fixed_dims = ['Analytic', 'Clout', 'Authentic', 'Tone', 'power']
    
    results = []
    country_sample_sizes = {}  # Store sample sizes for each country
    
    for country in countries:
        if country == 'All Countries':
            country_data = data.copy()
        else:
            country_data = data[data['Country'] == country].copy()
        
        filtered = country_data[country_data[covariate_col].isin(categories)].copy()
        
        # Store sample sizes for this country
        country_sample_sizes[country] = filtered[covariate_col].value_counts()
        
        # Z-score normalize
        benchmark_lookup = benchmarks.set_index('Dimension')
        for dim in dimensions:
            if dim in benchmark_lookup.index:
                mean = benchmark_lookup.loc[dim, 'Mean']
                std = benchmark_lookup.loc[dim, 'Std']
                filtered[f'{dim}_z'] = (filtered[dim] - mean) / std
        
        means = filtered.groupby(covariate_col)[[f'{d}_z' for d in dimensions]].mean()
        means.columns = [c.replace('_z', '') for c in means.columns]
        
        # Calculate difference
        if len(categories) == 2:
            diff = means.iloc[0] - means.iloc[1]
            results.append(diff)
   
    # Create dataframe with all countries
    all_results = pd.DataFrame(results, index=countries).T
    
    # Order by absolute difference and select top 5 non-fixed dimensions
    non_fixed_dims = [d for d in all_results.index if d not in fixed_dims]
    avg_abs = all_results.loc[non_fixed_dims].abs().mean(axis=1).sort_values(ascending=False)
    top_5_other = avg_abs.head(5).index.tolist()
    
    # Combine fixed + top 5
    selected_dims = fixed_dims + top_5_other
    all_results = all_results.loc[selected_dims]
    
    fig, ax = plt.subplots(figsize=(12, 8))
    sns.heatmap(all_results, annot=True, fmt='.2f', cmap='RdBu_r', center=0, 
                vmin=-1, vmax=1, ax=ax, square=False,
                cbar_kws={'label': 'Z-Score Diff (Coalition - Opposition)' if categories[0] == 'Coalition'
                         else 'Z-Score Diff (Female - Male)'},
                annot_kws={'color': 'black'})
    
    # Add separator line after first 5 dimensions
    ax.axhline(y=5, color='black', linewidth=2)
    
    ax.set_title(title, fontweight='bold')
    
    # Update column labels with sample sizes - 2 lines for better readability
    col_labels = []
    for country in countries:
        sizes = country_sample_sizes[country]
        if categories[0] == 'Coalition':
            coal_n = sizes.get('Coalition', 0)
            opp_n = sizes.get('Opposition', 0)
            col_labels.append(f"{country}\nCoal={coal_n:,}\nOpp={opp_n:,}")
        else:  # Gender
            m_n = sizes.get('M', 0)
            f_n = sizes.get('F', 0)
            col_labels.append(f"{country}\nM={m_n:,}\nF={f_n:,}")
    
    ax.set_xticklabels(col_labels, rotation=45, ha='right')
    
    plt.tight_layout()
    plt.savefig(os.path.join(liwc_dir, filename), dpi=300, bbox_inches='tight')
    plt.show()

# Coalition vs Opposition - 10 dimensions with separator
create_country_split_heatmap(LIWC_ALL, LIWC_benchmarks, 'Party_status', 
                             ['Coalition', 'Opposition'], KEY_LIWC_DIMENSIONS,
                             'Coalition vs Opposition: Top 10 Linguistic Differences', 
                             'liwc_party_status.png')

# Gender Differences - 10 dimensions with separator
create_country_split_heatmap(LIWC_ALL, LIWC_benchmarks, 'Speaker_gender', 
                             ['M', 'F'], KEY_LIWC_DIMENSIONS,
                             'Gender Differences: Top 10 Linguistic Dimensions', 
                             'liwc_gender.png')

# Political Orientation (Left-Right spectrum)
if 'Party_orientation' in LIWC_ALL.columns:
    # Filter out missing values
    orientation_data = LIWC_ALL[LIWC_ALL['Party_orientation'] != '-'].dropna(subset=['Party_orientation']).copy()
    
    # Define left-to-right spectrum order
    spectrum_order = [
        'Left',
        'Centre-left to left',
        'Centre-left',
        'Centre to centre-left',
        'Centre',
        'Centre to centre-right',
        'Centre-right',
        'Centre-right to right',
        'Right',
        'Right to far-right',
        'Far-right'
    ]
    
    # Fixed dimensions (always first 5)
    fixed_dims = ['Analytic', 'Clout', 'Authentic', 'Tone', 'power']
    
    # Z-score normalize all KEY_LIWC_DIMENSIONS
    benchmark_lookup = LIWC_benchmarks.set_index('Dimension')
    
    for dim in KEY_LIWC_DIMENSIONS:
        if dim in benchmark_lookup.index:
            mean = benchmark_lookup.loc[dim, 'Mean']
            std = benchmark_lookup.loc[dim, 'Std']
            orientation_data[f'{dim}_z'] = (orientation_data[dim] - mean) / std
    
    # Group by orientation and calculate means
    orientation_means = orientation_data.groupby('Party_orientation')[[f'{d}_z' for d in KEY_LIWC_DIMENSIONS]].mean()
    orientation_means.columns = [c.replace('_z', '') for c in orientation_means.columns]
    
    # Get sample sizes
    orientation_counts = orientation_data['Party_orientation'].value_counts()
    
    # Filter to only orientations that exist in data and order by spectrum
    available_orientations = [o for o in spectrum_order if o in orientation_means.index]
    orientation_means = orientation_means.loc[available_orientations]
    
    # Select top 5 non-fixed dimensions by highest absolute z-score
    non_fixed_dims = [d for d in orientation_means.columns if d not in fixed_dims]
    avg_abs_z = orientation_means[non_fixed_dims].abs().mean(axis=0).sort_values(ascending=False)
    top_5_other = avg_abs_z.head(5).index.tolist()
    
    # Combine fixed + top 5
    selected_dims = fixed_dims + top_5_other
    orientation_means = orientation_means[selected_dims]
    
    # Create labels with sample sizes
    orientation_labels = [f"{orient}\n(N={orientation_counts[orient]:,})" 
                         for orient in orientation_means.index]
    
    fig, ax = plt.subplots(figsize=(16, 10))
    sns.heatmap(orientation_means.T, annot=True, fmt='.2f', cmap='RdBu_r', 
               center=0, vmin=-1.5, vmax=1.5, ax=ax, square=False,
               cbar_kws={'label': 'Z-Score'},
               annot_kws={'color': 'black', 'size': 9})
    
    # Add separator line after first 5 dimensions
    ax.axhline(y=5, color='black', linewidth=2)
    
    ax.set_xticklabels(orientation_labels, rotation=45, ha='right')
    ax.set_title('Political Orientation Differences: Linguistic Dimensions\n(Left to Right Spectrum)', 
                fontweight='bold', fontsize=14)
    ax.set_xlabel('Political Orientation', fontweight='bold')
    ax.set_ylabel('LIWC Dimension', fontweight='bold')
    
    plt.tight_layout()
    plt.savefig(os.path.join(liwc_dir, 'political_orientation.png'), 
                dpi=300, bbox_inches='tight')
    plt.show()
    
    print(f"✅ Political orientation analysis complete: {len(available_orientations)} categories, {len(orientation_data):,} speeches")
else:
    print("⚠️ Party_orientation column not found in dataset")

# Age Groups
if 'Speaker_age' in LIWC_ALL.columns:
    age_data = LIWC_ALL.dropna(subset=['Speaker_age']).copy()
    age_data['Age_Group'] = pd.cut(age_data['Speaker_age'], 
                                     bins=[0, 35, 50, 65, 100], 
                                     labels=['<35', '35-50', '51-65', '66+'])
    
    age_dims = ['Analytic', 'Authentic', 'Clout', 'Tone', 'power', 
                'ipron', 'politic', 'certitude', 'money', 'Social']
    
    # Get sample sizes
    age_counts = age_data['Age_Group'].value_counts()
    
    # Z-score and group
    benchmark_lookup = LIWC_benchmarks.set_index('Dimension')
    for dim in age_dims:
        if dim in benchmark_lookup.index:
            mean = benchmark_lookup.loc[dim, 'Mean']
            std = benchmark_lookup.loc[dim, 'Std']
            age_data[f'{dim}_z'] = (age_data[dim] - mean) / std
    
    means = age_data.groupby('Age_Group')[[f'{d}_z' for d in age_dims]].mean()
    means.columns = [c.replace('_z', '') for c in means.columns]
    
    # Create labels with sample sizes
    age_labels = [f"{age}\n(N={age_counts[age]:,})" for age in means.index]
    
    fig, ax = plt.subplots(figsize=(10, 8))
    sns.heatmap(means.T, annot=True, fmt='.2f', cmap='RdBu_r', center=0, 
                vmin=-2, vmax=2, ax=ax, square=False,
                cbar_kws={'label': 'Z-Score'},
                annot_kws={'color': 'black'})
    
    # Add separator line after first 5 dimensions
    ax.axhline(y=5, color='black', linewidth=2)
    
    ax.set_xticklabels(age_labels, rotation=0)
    ax.set_title('Age Group Differences: Linguistic Dimensions', fontweight='bold')
    ax.set_xlabel('Age Group', fontweight='bold')
    ax.set_ylabel('LIWC Dimension', fontweight='bold')
    
    plt.tight_layout()
    plt.savefig(os.path.join(liwc_dir, 'liwc_age_groups.png'), 
                dpi=300, bbox_inches='tight')
    plt.show()

print("✅ All covariate analyses complete")
```

## Temporal Evolution Analysis

```python
def create_temporal_plot(data, benchmarks, dimensions, countries, title, filename, add_events=True):
    """Temporal evolution with shared x-axis, background shading, and event markers"""
    benchmark_lookup = benchmarks.set_index('Dimension')
    temporal_data = data[data['Date'].notna()].copy()
    
    for dim in dimensions:
        mean = benchmark_lookup.loc[dim, 'Mean']
        std = benchmark_lookup.loc[dim, 'Std']
        temporal_data[f'{dim}_z'] = (temporal_data[dim] - mean) / std
    
    # Determine overall date range from all countries
    min_date = temporal_data['Date'].min()
    max_date = temporal_data['Date'].max()
    
    # Key events to annotate
    events = {
        'Austria': [
            ('2008-09-15', 'Financial\nCrisis'),
            ('2015-09-01', 'Refugee\nCrisis'),
            ('2020-03-16', 'COVID-19'),
            ('2022-02-24', 'Ukraine\nWar')
        ],
        'Croatia': [
            ('2008-09-15', 'Financial\nCrisis'),
            ('2013-07-01', 'EU\nAdmission'),
            ('2015-09-01', 'Refugee\nCrisis'),
            ('2020-03-16', 'COVID-19'),
            ('2022-02-24', 'Ukraine\nWar')
        ],
        'Great Britain': [
            ('2015-09-01', 'Refugee\nCrisis'),
            ('2020-03-16', 'COVID-19'),
            ('2022-02-24', 'Ukraine\nWar')
        ]
    }
    
    # Political periods - simplified to 2 colors alternating at elections
    periods = {
        'Austria': [
            ('1996-01-01', '2000-02-04', '#8DB4E2'),  # Color 1
            ('2000-02-04', '2007-01-11', '#C4D79B'),  # Color 2 (election)
            ('2007-01-11', '2017-12-18', '#8DB4E2'),  # Color 1 (election)
            ('2017-12-18', '2019-06-03', '#C4D79B'),  # Color 2 (election)
            ('2019-06-03', '2020-01-07', '#8DB4E2'),  # Color 1 (election)
            ('2020-01-07', '2022-12-31', '#C4D79B')   # Color 2 (election)
        ],
        'Croatia': [
            ('2004-01-01', '2011-12-23', '#8DB4E2'),  # Color 1
            ('2011-12-23', '2016-01-22', '#C4D79B'),  # Color 2 (election)
            ('2016-01-22', '2022-12-31', '#8DB4E2')   # Color 1 (election)
        ],
        'Great Britain': [
            ('2015-01-01', '2016-07-13', '#8DB4E2'),  # Color 1
            ('2016-07-13', '2019-07-24', '#C4D79B'),  # Color 2 (election)
            ('2019-07-24', '2022-12-31', '#8DB4E2')   # Color 1 (election)
        ]
    }
    
    fig, axes = plt.subplots(3, 1, figsize=(20, 14), sharex=True)
    fig.suptitle(title, fontsize=16, fontweight='bold', y=0.995);
    
    for idx, country in enumerate(countries):
        ax = axes[idx]
        
        # Add background shading for political periods
        if add_events and country in periods:
            for start, end, color in periods[country]:
                start_date = pd.to_datetime(start)
                end_date = pd.to_datetime(end)
                ax.axvspan(start_date, end_date, alpha=0.15, color=color, zorder=0)
        
        # Plot data
        country_data = temporal_data[temporal_data['Country'] == country].copy()
        country_data['YearMonth'] = country_data['Date'].dt.to_period('M')
        
        monthly = country_data.groupby('YearMonth')[[f'{d}_z' for d in dimensions]].mean()
        monthly.columns = [c.replace('_z', '') for c in monthly.columns]
        monthly.index = monthly.index.to_timestamp()
        monthly_smooth = monthly.rolling(window=3, center=True).mean()
        
        for dim in monthly_smooth.columns:
            ax.plot(monthly_smooth.index, monthly_smooth[dim], linewidth=2.5, label=dim, alpha=0.9, zorder=2)
        
        # Add event annotations
        if add_events and country in events:
            for event_date, event_label in events[country]:
                event_dt = pd.to_datetime(event_date)
                if min_date <= event_dt <= max_date:
                    # Draw vertical line
                    ax.axvline(x=event_dt, color='red', linestyle='--', 
                              linewidth=1.5, alpha=0.6, zorder=1)
                    
                    # Add text annotation at top of plot
                    ylim = ax.get_ylim()
                    y_pos = ylim[1] * 0.95
                    ax.text(event_dt, y_pos, event_label, 
                           rotation=90, verticalalignment='top', horizontalalignment='right',
                           fontsize=16, color='black', fontweight='bold', alpha=1)
        
        ax.axhline(y=0, color='gray', linestyle='-', alpha=0.5, linewidth=1.5, zorder=1)
        ax.set_title(country, fontweight='bold', loc='left', fontsize=13)
        ax.set_ylabel('Z-Score (3-month avg)', fontweight='bold')
        ax.set_xlabel('Year', fontweight='bold', fontsize=12)
        ax.grid(True, alpha=0.2, linestyle=':', zorder=0)
        ax.set_xlim(min_date, max_date)
        ax.tick_params(labelbottom=True)
        
        plt.setp(ax.xaxis.get_majorticklabels(), rotation=0)
        
        # Legend - only dimension lines, no events
        if idx == 0:
            handles, labels = ax.get_legend_handles_labels()
            # Filter to only include dimension lines (not event markers)
            dim_handles = [h for h, l in zip(handles, labels) if l in dimensions]
            dim_labels = [l for l in labels if l in dimensions]
            ax.legend(dim_handles, dim_labels, 
                     loc='upper left', ncol=len(dimensions), fontsize=9, framealpha=0.9)
    
    plt.tight_layout(rect=[0, 0.01, 1, 0.99])
    plt.savefig(os.path.join(temporal_liwc_dir, filename), dpi=300, bbox_inches='tight')
    plt.show()

# Summary Variables
create_temporal_plot(LIWC_ALL, LIWC_benchmarks, ['Analytic', 'Authentic', 'Clout', 'Tone'],
                    ['Austria', 'Croatia', 'Great Britain'], 
                    'Temporal Evolution: Summary Variables', 'liwc_temporal_summary.png')

# Political Language
create_temporal_plot(LIWC_ALL, LIWC_benchmarks, ['politic', 'power', 'moral', 'money'],
                    ['Austria', 'Croatia', 'Great Britain'],
                    'Temporal Evolution: Political Language', 'liwc_temporal_political.png')

# Pronouns
create_temporal_plot(LIWC_ALL, LIWC_benchmarks, ['i', 'we', 'you', 'they'],
                    ['Austria', 'Croatia', 'Great Britain'],
                    'Temporal Evolution: Pronoun Usage', 'liwc_temporal_pronouns.png')

# Cognitive Processes
create_temporal_plot(LIWC_ALL, LIWC_benchmarks, ['cogproc', 'insight', 'cause', 'certitude'],
                    ['Austria', 'Croatia', 'Great Britain'],
                    'Temporal Evolution: Cognitive Processes', 'liwc_temporal_cognitive.png')

# Temporal Focus
create_temporal_plot(LIWC_ALL, LIWC_benchmarks, ['focuspast', 'focuspresent', 'focusfuture'],
                    ['Austria', 'Croatia', 'Great Britain'],
                    'Temporal Evolution: Time Orientation', 'liwc_temporal_time.png')

# Affect and Tone
create_temporal_plot(LIWC_ALL, LIWC_benchmarks, ['Affect', 'tone_pos', 'tone_neg'],
                    ['Austria', 'Croatia', 'Great Britain'],
                    'Temporal Evolution: Emotional Language', 'liwc_temporal_affect.png')

print("✅ All temporal plots created with events and political periods")
```

## Topic Prevalence Over Time

```python
def create_topic_prevalence_plot(data, topics, countries, title, filename, add_events=True):
    """Create topic prevalence over time with event markers and political periods"""
    temporal_data = data[data['Date'].notna()].copy()
    
    # Determine overall date range
    min_date = temporal_data['Date'].min()
    max_date = temporal_data['Date'].max()
    
    # Key events to annotate
    events = {
        'Austria': [
            ('2008-09-15', 'Financial\nCrisis'),
            ('2015-09-01', 'Refugee\nCrisis'),
            ('2020-03-16', 'COVID-19'),
            ('2022-02-24', 'Ukraine\nWar')
        ],
        'Croatia': [
            ('2008-09-15', 'Financial\nCrisis'),
            ('2013-07-01', 'EU\nAccession'),
            ('2015-09-01', 'Refugee\nCrisis'),
            ('2020-03-16', 'COVID-19'),
            ('2022-02-24', 'Ukraine\nWar')
        ],
        'Great Britain': [
            ('2015-09-01', 'Refugee\nCrisis'),
            ('2020-03-16', 'COVID-19'),
            ('2022-02-24', 'Ukraine\nWar')
        ]
    }
    
    # Political periods
    periods = {
        'Austria': [
            ('1996-01-01', '2000-02-04', '#8DB4E2'),
            ('2000-02-04', '2007-01-11', '#C4D79B'),
            ('2007-01-11', '2017-12-18', '#8DB4E2'),
            ('2017-12-18', '2019-06-03', '#C4D79B'),
            ('2019-06-03', '2020-01-07', '#8DB4E2'),
            ('2020-01-07', '2022-12-31', '#C4D79B')
        ],
        'Croatia': [
            ('2004-01-01', '2011-12-23', '#8DB4E2'),
            ('2011-12-23', '2016-01-22', '#C4D79B'),
            ('2016-01-22', '2022-12-31', '#8DB4E2')
        ],
        'Great Britain': [
            ('2015-01-01', '2016-07-13', '#8DB4E2'),
            ('2016-07-13', '2019-07-24', '#C4D79B'),
            ('2019-07-24', '2022-12-31', '#8DB4E2')
        ]
    }
    
    fig, axes = plt.subplots(3, 1, figsize=(20, 14), sharex=True)
    fig.suptitle(title, fontsize=16, fontweight='bold', y=0.995);
    
    for idx, country in enumerate(countries):
        ax = axes[idx]
        
        # Add background shading for political periods
        if add_events and country in periods:
            for start, end, color in periods[country]:
                start_date = pd.to_datetime(start)
                end_date = pd.to_datetime(end)
                ax.axvspan(start_date, end_date, alpha=0.15, color=color, zorder=0)
        
        # Filter country data
        country_data = temporal_data[temporal_data['Country'] == country].copy()
        country_data['YearMonth'] = country_data['Date'].dt.to_period('M')
        
        # Calculate topic prevalence as percentage of ALL speeches
        monthly_totals = country_data.groupby('YearMonth').size()
        
        # Filter for selected topics AFTER calculating totals
        topic_data = country_data[country_data['Our_Topic'].isin(topics)]
        topic_monthly = topic_data.groupby(['YearMonth', 'Our_Topic']).size().unstack(fill_value=0)
        
        # Now calculate prevalence against ALL speeches
        topic_prevalence = (topic_monthly.div(monthly_totals, axis=0) * 100)
        
        # Convert to timestamp for plotting
        topic_prevalence.index = topic_prevalence.index.to_timestamp()
        
        # FILL MISSING MONTHS: Create complete date range and reindex
        date_range = pd.date_range(start=topic_prevalence.index.min(), 
                                   end=topic_prevalence.index.max(), 
                                   freq='MS')  # Month Start frequency
        topic_prevalence = topic_prevalence.reindex(date_range)
        
        # Fill gaps with forward fill (or use interpolate for smoother lines)
        topic_prevalence = topic_prevalence.fillna(0)  # Use 0 for months with no speeches
        # Alternatively: topic_prevalence = topic_prevalence.interpolate(method='linear')
        
        # Apply 3-month rolling average
        topic_prevalence_smooth = topic_prevalence.rolling(window=3, center=True).mean()
        
        # Plot each topic
        for topic in topics:
            if topic in topic_prevalence_smooth.columns:
                ax.plot(topic_prevalence_smooth.index, topic_prevalence_smooth[topic], 
                       linewidth=2.5, label=topic, alpha=0.9, zorder=2)
        
        # Add event annotations
        if add_events and country in events:
            for event_date, event_label in events[country]:
                event_dt = pd.to_datetime(event_date)
                if min_date <= event_dt <= max_date:
                    ax.axvline(x=event_dt, color='red', linestyle='--', 
                              linewidth=1.5, alpha=0.6, zorder=1)
                    
                    ylim = ax.get_ylim()
                    y_pos = ylim[1] * 0.95
                    ax.text(event_dt, y_pos, event_label, 
                           rotation=90, verticalalignment='top', horizontalalignment='right',
                           fontsize=16, color='black', fontweight='bold', alpha=1)
        
        ax.set_title(f'{country} (1996-2022)' if country == 'Austria' 
                    else f'{country} (2003-2022)' if country == 'Croatia'
                    else f'{country} (2015-2022)', 
                    fontweight='bold', loc='left', fontsize=13)
        ax.set_ylabel('Topic Prevalence', fontweight='bold')
        ax.set_xlabel('Year', fontweight='bold', fontsize=12)
        ax.grid(True, alpha=0.2, linestyle=':', zorder=0)
        ax.set_xlim(min_date, max_date)
        ax.tick_params(labelbottom=True)
        
        plt.setp(ax.xaxis.get_majorticklabels(), rotation=0)
        
        # Legend
        if idx == 0:
            ax.legend(loc='upper left', ncol=len(topics), fontsize=9, framealpha=0.9)
    
    plt.tight_layout(rect=[0, 0.01, 1, 0.99])
    plt.savefig(os.path.join(temporal_topic_dir, filename), dpi=300, bbox_inches='tight')
    plt.show()

# Economic Topics (Macroeconomics & Health as shown in your image)
create_topic_prevalence_plot(LIWC_ALL, ['Macroeconomics', 'Health'],
                            ['Austria', 'Croatia', 'Great Britain'],
                            'Temporal Evolution of Economic Topics',
                            'topic_prevalence_economic.png')

# Social Topics
create_topic_prevalence_plot(LIWC_ALL, ['Social Welfare', 'Education'],
                            ['Austria', 'Croatia', 'Great Britain'],
                            'Temporal Evolution of Social Policy Topics',
                            'topic_prevalence_social.png')

# Security & Foreign Affairs
create_topic_prevalence_plot(LIWC_ALL, ['Law and Crime', 'International Affairs'],
                            ['Austria', 'Croatia', 'Great Britain'],
                            'Temporal Evolution of Security Topics',
                            'topic_prevalence_security.png')


print("✅ Topic prevalence plots created")
```

## Party-Level Temporal Analysis

```python
def create_party_temporal_plot(data, benchmarks, dimension, country, parties, title, filename):
    """Create party-level temporal evolution with coalition/opposition status and period means"""
    
    # Filter for specific country and parties
    party_data = data[(data['Country'] == country) & (data['Speaker_party'].isin(parties))].copy()
    party_data = party_data[party_data['Date'].notna()]
    
    # IMPORTANT: Filter out speeches with missing or invalid Party_status
    party_data = party_data[party_data['Party_status'].isin(['Coalition', 'Opposition'])].copy()
    
    # Z-score normalize
    benchmark_lookup = benchmarks.set_index('Dimension')
    if dimension in benchmark_lookup.index:
        mean = benchmark_lookup.loc[dimension, 'Mean']
        std = benchmark_lookup.loc[dimension, 'Std']
        party_data[f'{dimension}_z'] = (party_data[dimension] - mean) / std
    
    party_data['YearMonth'] = party_data['Date'].dt.to_period('M')
    party_data['Year'] = party_data['Date'].dt.year
    
    # Get date range
    min_date = party_data['Date'].min()
    max_date = party_data['Date'].max()
    
    fig, axes = plt.subplots(len(parties), 1, figsize=(20, 4*len(parties)), sharex=True)
    if len(parties) == 1:
        axes = [axes]
    
    fig.suptitle(title, fontsize=16, fontweight='bold', y=0.995)
    
    for idx, party in enumerate(parties):
        ax = axes[idx]
        
        # Filter party data
        single_party = party_data[party_data['Speaker_party'] == party].copy()
        
        if len(single_party) == 0:
            print(f"⚠️ Warning: No valid data for {party}")
            continue
        
        # Get status by YEAR (most stable) - this gives us government periods
        yearly_status = single_party.groupby('Year')['Party_status'].agg(
            lambda x: x.mode()[0] if len(x.mode()) > 0 else x.iloc[0]
        ).reset_index()
        
        # Create continuous date ranges for background shading
        yearly_status = yearly_status.sort_values('Year')
        yearly_status['status_change'] = (yearly_status['Party_status'] != yearly_status['Party_status'].shift()).cumsum()
        
        # Background shading by coalition/opposition status WITH MEAN ANNOTATIONS
        for group_id in yearly_status['status_change'].unique():
            group_data = yearly_status[yearly_status['status_change'] == group_id]
            status = group_data['Party_status'].iloc[0]
            start_year = group_data['Year'].min()
            end_year = group_data['Year'].max()
            
            # Convert years to dates
            start_date = pd.Timestamp(f'{start_year}-01-01')
            end_date = pd.Timestamp(f'{end_year}-12-31')
            
            color = '#C4D79B' if status == 'Coalition' else '#FFB6C1'
            ax.axvspan(start_date, end_date, alpha=0.25, color=color, zorder=0)
        
        # Plot smoothed line
        monthly_agg = single_party.groupby('YearMonth')[f'{dimension}_z'].mean()
        monthly_agg.index = monthly_agg.index.to_timestamp()
        monthly_smooth = monthly_agg.rolling(window=3, center=True).mean()
        
        ax.plot(monthly_smooth.index, monthly_smooth.values, linewidth=2.5, 
               color='navy', alpha=0.9, zorder=2)
        
        # Add horizontal line at 0
        ax.axhline(y=0, color='gray', linestyle='-', alpha=0.5, linewidth=1.5, zorder=1)
        
        # SET CONSISTENT Y-AXIS LIMITS (standardized range for better visual comparison)
        ax.set_ylim(-1.5, 1.5)  # Standard range for z-scores
        
        # NOW calculate period means and add annotations (after setting ylim)
        for group_id in yearly_status['status_change'].unique():
            group_data = yearly_status[yearly_status['status_change'] == group_id]
            status = group_data['Party_status'].iloc[0]
            start_year = group_data['Year'].min()
            end_year = group_data['Year'].max()
            
            # Convert years to dates
            start_date = pd.Timestamp(f'{start_year}-01-01')
            end_date = pd.Timestamp(f'{end_year}-12-31')
            
            # Calculate mean for this period
            period_data = single_party[
                (single_party['Year'] >= start_year) & 
                (single_party['Year'] <= end_year)
            ]
            period_mean = period_data[f'{dimension}_z'].mean()
            
            # Add mean annotation in the middle of the period
            mid_date = start_date + (end_date - start_date) / 2
            
            # Position text at top of plot area (now consistent across all plots)
            y_pos = 1.4  # Fixed position near top
            
            # Add text with mean value
            status_label = 'Coal' if status == 'Coalition' else 'Opp'
            color = '#C4D79B' if status == 'Coalition' else '#FFB6C1'
            ax.text(mid_date, y_pos, f'{status_label}\nμ={period_mean:.2f}',
                   fontsize=13, fontweight='bold',
                   color='darkgreen' if status == 'Coalition' else 'darkred',
                   horizontalalignment='center', verticalalignment='top',
                   bbox=dict(boxstyle='round,pad=0.4', facecolor='white', 
                            edgecolor=color, alpha=0.95, linewidth=2),
                   zorder=3)
        
        # Get coalition/opposition time info (from FILTERED data)
        coalition_months = single_party[single_party['Party_status'] == 'Coalition']['YearMonth'].nunique()
        opposition_months = single_party[single_party['Party_status'] == 'Opposition']['YearMonth'].nunique()
        
        # Calculate overall means for coalition and opposition periods
        coalition_mean = single_party[single_party['Party_status'] == 'Coalition'][f'{dimension}_z'].mean()
        opposition_mean = single_party[single_party['Party_status'] == 'Opposition'][f'{dimension}_z'].mean()
        
        ax.set_title(f'{party} | Coalition: {coalition_months}mo (μ={coalition_mean:.2f}) | Opposition: {opposition_months}mo (μ={opposition_mean:.2f})', 
                    fontweight='bold', loc='left', fontsize=13)
        ax.set_ylabel(f'{dimension} Z-Score\n(3-month avg)', fontweight='bold')
        ax.grid(True, alpha=0.2, linestyle=':', zorder=0)
        ax.set_xlim(min_date, max_date)
        ax.tick_params(labelbottom=True)
        
        plt.setp(ax.xaxis.get_majorticklabels(), rotation=0)
    
    axes[-1].set_xlabel('Year', fontweight='bold', fontsize=12)
    
    # Add legend
    # from matplotlib.patches import Patch
    # legend_elements = [
    #     Patch(facecolor='#C4D79B', alpha=0.3, label='Coalition'),
    #     Patch(facecolor='#FFB6C1', alpha=0.3, label='Opposition')
    # ]
    # axes[0].legend(handles=legend_elements, loc='upper right', fontsize=10, framealpha=0.9)
    
    plt.tight_layout(rect=[0, 0.01, 1, 0.99])
    plt.savefig(os.path.join(temporal_party_dir, filename), dpi=300, bbox_inches='tight')
    plt.show()

# Austria - All dimensions
austria_parties = ['SPÖ', 'ÖVP', 'FPÖ', 'Grüne']

create_party_temporal_plot(LIWC_ALL, LIWC_benchmarks, 'Analytic', 'Austria', 
                           austria_parties,
                           'Austria Party Analytic Evolution with Coalition/Opposition Status',
                           'party_analytic_austria.png')

create_party_temporal_plot(LIWC_ALL, LIWC_benchmarks, 'Authentic', 'Austria', 
                           austria_parties,
                           'Austria Party Authentic Evolution with Coalition/Opposition Status',
                           'party_authentic_austria.png')

create_party_temporal_plot(LIWC_ALL, LIWC_benchmarks, 'emo_anger', 'Austria', 
                           austria_parties,
                           'Austria Party Anger Evolution with Coalition/Opposition Status',
                           'party_anger_austria.png')

create_party_temporal_plot(LIWC_ALL, LIWC_benchmarks, 'emo_anx', 'Austria', 
                           austria_parties,
                           'Austria Party Anxiety Evolution with Coalition/Opposition Status',
                           'party_anxiety_austria.png')

create_party_temporal_plot(LIWC_ALL, LIWC_benchmarks, 'emo_sad', 'Austria', 
                           austria_parties,
                           'Austria Party Sadness Evolution with Coalition/Opposition Status',
                           'party_sadness_austria.png')

create_party_temporal_plot(LIWC_ALL, LIWC_benchmarks, 'moral', 'Austria', 
                           austria_parties,
                           'Austria Party Moral Language Evolution with Coalition/Opposition Status',
                           'party_moral_austria.png')

create_party_temporal_plot(LIWC_ALL, LIWC_benchmarks, 'Tone', 'Austria', 
                           austria_parties,
                           'Austria Party Tone Evolution with Coalition/Opposition Status',
                           'party_tone_austria.png')

# Croatia - Same dimensions, expanded parties
croatia_parties = ['HDZ', 'SDP', 'MOST', 'HNS', 'HSS']

create_party_temporal_plot(LIWC_ALL, LIWC_benchmarks, 'Analytic', 'Croatia', 
                           croatia_parties,
                           'Croatia Party Analytic Evolution with Coalition/Opposition Status',
                           'party_analytic_croatia.png')

create_party_temporal_plot(LIWC_ALL, LIWC_benchmarks, 'Authentic', 'Croatia', 
                           croatia_parties,
                           'Croatia Party Authentic Evolution with Coalition/Opposition Status',
                           'party_authentic_croatia.png')

create_party_temporal_plot(LIWC_ALL, LIWC_benchmarks, 'emo_anger', 'Croatia', 
                           croatia_parties,
                           'Croatia Party Anger Evolution with Coalition/Opposition Status',
                           'party_anger_croatia.png')

create_party_temporal_plot(LIWC_ALL, LIWC_benchmarks, 'emo_anx', 'Croatia', 
                           croatia_parties,
                           'Croatia Party Anxiety Evolution with Coalition/Opposition Status',
                           'party_anxiety_croatia.png')

create_party_temporal_plot(LIWC_ALL, LIWC_benchmarks, 'emo_sad', 'Croatia', 
                           croatia_parties,
                           'Croatia Party Sadness Evolution with Coalition/Opposition Status',
                           'party_sadness_croatia.png')

create_party_temporal_plot(LIWC_ALL, LIWC_benchmarks, 'moral', 'Croatia', 
                           croatia_parties,
                           'Croatia Party Moral Language Evolution with Coalition/Opposition Status',
                           'party_moral_croatia.png')

create_party_temporal_plot(LIWC_ALL, LIWC_benchmarks, 'Tone', 'Croatia', 
                           croatia_parties,
                           'Croatia Party Tone Evolution with Coalition/Opposition Status',
                           'party_tone_croatia.png')
```

```python
import pandas as pd
import numpy as np

def calculate_token_statistics(data, text_column='Text_English', segment_id_column='Segment_ID_English', token_window=8192):
    """Calculate token statistics at speech and segment level using actual segment groupings"""
    
    results = []
    
    for country in ['Austria', 'Croatia', 'Great Britain']:
        country_data = data[data['Country'] == country].copy()
        
        # Speech-level statistics
        country_data['token_count'] = country_data[text_column].apply(lambda x: len(str(x).split()))
        
        speech_exceeding = (country_data['token_count'] > token_window).sum()
        speech_total = len(country_data)
        speech_pct = (speech_exceeding / speech_total * 100) if speech_total > 0 else 0
        
        # Segment-level statistics using Segment_ID_English
        # Group by segment ID and sum tokens for each segment
        if segment_id_column in country_data.columns:
            segment_tokens = country_data.groupby(segment_id_column)['token_count'].sum()
            segment_count = len(segment_tokens)
            segment_exceeding = (segment_tokens > token_window).sum()
        else:
            # Fallback if column doesn't exist
            segment_count = speech_total
            segment_exceeding = speech_exceeding
        
        segment_pct = (segment_exceeding / segment_count * 100) if segment_count > 0 else 0
        
        # Average speeches per segment
        avg_speeches_per_segment = speech_total / segment_count if segment_count > 0 else 0
        
        results.append({
            'Country': country,
            'Speech (%)': round(speech_pct, 2),
            'Segment (%)': round(segment_pct, 2),
            'Avg Speeches per Segment': round(avg_speeches_per_segment, 2),
            'Total Speeches': speech_total,
            'Total Segments': segment_count
        })
    
    # Calculate totals
    total_speeches = sum(r['Total Speeches'] for r in results)
    total_segments = sum(r['Total Segments'] for r in results)
    total_speech_pct = sum(r['Speech (%)'] * r['Total Speeches'] for r in results) / total_speeches
    total_segment_pct = sum(r['Segment (%)'] * r['Total Segments'] for r in results) / total_segments
    total_avg_speeches = total_speeches / total_segments
    
    results.append({
        'Country': 'Total',
        'Speech (%)': round(total_speech_pct, 2),
        'Segment (%)': round(total_segment_pct, 2),
        'Avg Speeches per Segment': round(total_avg_speeches, 2),
        'Total Speeches': total_speeches,
        'Total Segments': total_segments
    })
    
    return pd.DataFrame(results)

# Calculate statistics
token_stats = calculate_token_statistics(LIWC_ALL, text_column='Text_English', 
                                        segment_id_column='Segment_ID_English', 
                                        token_window=8192)

print("\n📊 Token Window Statistics (8,192-token threshold):")
print("=" * 80)
display(token_stats[['Country', 'Speech (%)', 'Segment (%)', 'Avg Speeches per Segment']])
```
